```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import load_model
```

```python
matches_for_predict = pd.read_csv('matches_for_predict.csv')
futureM = pd.read_csv('Futures.csv')
model = load_model("FinalData.h5")
```

```python
predictions = model.predict(matches_for_predict)
for index, predict in enumerate(predictions):
    print(f'{futureM.loc[index, "Team"]} VS {futureM.loc[index, "Opponent"]}')
    if np.argmax(predict) == 1:
        print("***** Draw *****")
    elif np.argmax(predict) == 0:
        if futureM['Venue'].iloc[index] == 'Home':
            print(
                f"Home team wins. ***** {futureM['Team'].iloc[index]} *****")
        elif futureM['Venue'].iloc[index] == 'Away':
            print(
                f"Away team wins. ***** {futureM['Team'].iloc[index]} *****")
    else:
        if futureM['Venue'].iloc[index] == 'Home':
            print(
                f"Away team wins. ***** {futureM['Opponent'].iloc[index]} *****")
        elif futureM['Venue'].iloc[index] == 'Away':
            print(
                f"Away team wins. ***** {futureM['Team'].iloc[index]} *****")
    # print(predict)
    print('---------------------------------------------------------')
    print('---------------------------------------------------------')
```
