```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import load_model
```

```python
matches_for_predict = pd.read_csv('matches_for_predict.csv')
model = load_model("Final_data.h5")
```

```python
predictions = model.predict(matches_for_predict)
print('If result = 1 it means first team wins, if result = 2 it means second team wins, if result = 0 it means the full-time result is Draw.')
for index, predict in enumerate(predictions):
    print(f'The prediction between {matches_for_predict["Team"][index]} and {matches_for_predict["Opponent"][index]} is {np.argmax(predict)}')
    
```
